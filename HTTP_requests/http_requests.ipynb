```python
import requests

# Scope to specific providers with providers.
url = "https://api.frankfurter.dev/v2/rates?providers=ecb"

response = requests.get(url)
```

#### Status Code
##### 1xx -> Informational (The request was received and the server is continuing to process it.)
##### 2xx -> Success (200 OK)
##### 3xx -> Redirection (301 Moved Permanently)
##### 4xx -> Client Error (404 Not Found)
##### 5xx -> Server Error (500 Internal Server Error)

```python
try:
    response = requests.get("https://api.frankfurter.dev/v2/rates?providers=ecb")

    if response.status_code == 200:
        data = response.json()
        # print("Rate:", data[4])
        for item in data:
            print("Rate:", item['rate'])
    else:
        print("Something went wrong. Status code:", response.status_code)

except Exception as e:
    print("Could not reach the API: ", e)
```

```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()

# 1. Fetch the specific EUR to USD rate from the ECB provider
response = requests.get(
    "https://api.frankfurter.dev/v2/rates",
    params={"providers": "ecb", "base": "EUR", "quotes": "USD"}
)
data = response.json()

# 2. Extract the rate value directly from the response array
rate_value = data[0]["rate"]

# 3. Create a clean, non-JSON text prompt
prompt = f"Convert 100 EUR to USD using the current exchange rate of {rate_value} and explain the calculation."

try:
    answer = client.models.generate_content(
        model="gemini-3.6-flash",  
        contents=prompt,
    )
except Exception as e:
    answer = f"Something went wrong. {e}"

print(answer)
```
