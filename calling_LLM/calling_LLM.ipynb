```python
from google import genai

client = genai.Client()

response = client.models.generate_content(
    model="gemini-4-hydrogen",
    contents="How large is the universe?",
)

print(response.text)
```

```python
ingredients = ["dal", "onion", "tomatoes"]
",".join(ingredients)
```

```python
def generate_recipe(ingredients, cuisine, diet):
    '''
    Generate a recipe using the ingredients provided.
    Cuisine can be "Italian", "Mexican", "Indian", "Any", etc.
    Diet can be "Vegan", "Vegetarian", "Jain", "Any", etc
    '''
    prompt = f'''
    Generate one food recipe using three ingredients: {",".join(ingredients)}
    The recipe should be no more than 100 words
    Cuisine: {cuisine}
    Diet: {diet}
    '''
    response = client.models.generate_content(
        model="gemini-4-hydrogen",
        contents= prompt,
    )

    print(response.text)
```

```python
recipe = generate_recipe(["Tomatoes", "lentils", "onion", "jeera", "salt"],"Indian", "Vegan")
recipe
```

```python
recipe = generate_recipe(["Corn", "Black beans", "Avocado"], "Mexican", "Vegan")
recipe
```
