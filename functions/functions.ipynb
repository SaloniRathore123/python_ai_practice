#### FUNCTIONS
##### Function is a reusable block of code, which perform a specific task.

```python
def calculate_margin(revenue, expense):
    profit = revenue - expense
    margin = profit * 100 / revenue
    return margin
```

```python
calculate_margin(50, 10)
```

```python
calculate_margin(80, 60)
```

```python
def calculate_profit_and_margin(revenue, expense):
    profit = revenue - expense
    margin = profit * 100 / revenue
    return profit, margin
revenue = 50
expense = 10
profit, margin = calculate_profit_and_margin(revenue, expense)
profit
```

```python
response = calculate_profit_and_margin(revenue, expense)
type(response)
```

#### What is intrinsic datatype
##### An intrinsic data type (often called a built-in or primitive data type) is a basic data type that is natively defined and provided directly by a programming language or compiler, requiring no external libraries or custom definitions to use.
##### Intrinsic Types: Built directly into the language core (e.g., int, float, char).

#### Why does Python not use intrinsic data types?
##### Python does actually have built-in data types (such as int, float, str, and bool), but it does not require you to explicitly declare or use a static data type when creating a variable.
##### Works as a:
##### 1. Dynamic Typing
##### 2. Variable as a reference
##### 3. Design Philosophy

```python
def calculate_margin(revenue: float, expense: float) -> float:
    profit = revenue - expense
    margin = profit * 100 / revenue
    return margin
```

```python
def calculate_expense(exp1, exp2):
    return exp1 + exp2
calculate_expense(2,3)
```

#### How can I make a function that takes dynamic arguments?
##### Simply, through args.

```python
def cal_total_expense(*args):
    print(args)
    return sum(args)
cal_total_expense(2,3,4)
```

```python
def total_expense(rent, phone_bill):
    return rent + phone_bill
total_expense(rent=1000, phone_bill=30)
```

```python
def total_expenses(**kargs):
    total = 0
    for key, value in kargs.items():
        total += value
    return total
    
total_expenses(rent=1000, phone_bill=30)
```
