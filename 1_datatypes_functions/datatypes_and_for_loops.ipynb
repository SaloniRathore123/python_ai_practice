#### For Loop is better for printing the revenues and expenses one by one

```python
revenues = [50, 60, 55]
expenses = [10, 30, 35]
```

```python
for rev in revenues:
    print(rev)
    #print(f"Revenues: {rev}") formate string
```

```python
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    print(f"Profit Q{i+1}: {profit: .2f}")
    print(f"Margin Q{i+1}: {margin: .2f}")
```

#### You can add margin to another list through the append method

```python
margins = []
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    margins. append(margin)
margins
```

#### If I want to store both revenues and expenses in objects Q1, Q2, and so on. How can we do it?
#### A dictionary is the best option
#### A dictionary is like a hash map where you can access a value by key

```python
fin = {
    "revenue": 50, 
    "expense": 10
}
type(fin)
```

```python
profit = fin["revenue"] - fin["expense"]
profit
```

```python
fin['profit'] = fin["revenue"] - fin["expense"]
fin
```

```python
financials = [
    {"revenue": 50, "expense":10}, # Q1
    {"revenue": 60, "expense":30}, # Q2
    {"revenue": 55, "expense":35}  # Q3
]
financials
```

```python
for fin in financials:
    profit = fin["revenue"] - fin["expense"]
    fin['margin'] = profit * 100 / fin['revenue']
financials
```

##### Nested Dictionary

```python
financials = {
    "Q1": {"revenue": 50, "expense":10}, # Q1
    "Q2": {"revenue": 60, "expense":30}, # Q2
    "Q3": {"revenue": 55, "expense":35}  # Q3
}
for quarter, data in financials .items():
    print(f"Quarter: {quarter}")
    print(f"Data: {data}")
```

```python
for quarter, data in financials .items():
    margin = (data["revenue"] - data["expense"]) / data["revenue"] * 100
    print(f"{quarter}: {margin: .2f}%")
```

```python
financials = {
    "Q1": {"revenue": 50, "expense":10}, # Q1
    "Q2": {"revenue": 60, "expense":30}, # Q2
    "Q3": {"revenue": 55, "expense":35},  # Q3
    "Q4": {"revenue": -10, "expense":35}, # bug (revenue can't be negative)
}
```

##### Add a check to avoid this negative value.
##### Using an if condition

```python
for quarter, data in financials .items():
    if data["revenue"] < 0:
       print(f"Invalid revenue for {quarter}. Skipping")
       continue #skip it and go on to the next iteration
    margin = (data["revenue"] - data["expense"]) / data["revenue"] * 100
    print(f"{quarter}: {margin: .2f}%")    
```
