```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()

response = client.models.generate_content(
    model="gemini-3.6-flash",  # Switched to an alternative Flash model
    contents="Hi! my name is Saloni. Please remember it.",
)

print(response.text)
```

```python
response = client.models.generate_content(
    model="gemini-3.6-flash",  # Switched to an alternative Flash model
    contents="What is my name?",
)

print(response.text)
```

```python
conversation = [] # Chatbot memory
def chat(user_message):
    # 1. What the user said to memory
    conversation.append({"role": "user", "parts": [{"text": user_message}]}) # Convention in LLM

    # 2. Send the entire conversation so far
    try:
        response = client.models.generate_content(
            model="gemini-3.6-flash",  
            contents= conversation,
        )
        reply = response.text
    except Exception as e:
        return f"Sorry, something went wrong: ({e})."

    # 3. Add the AI's reply to memory so it remeber for next time
    conversation.append({"role": "user", "parts": [{"text": reply}]})
    return reply
```

```python
chat("Hi! My name is Saloni. Please remember it.")
```

```python
chat("What is my name?")
```

```python
for message in conversation:
    print(message['role'], "->", message['parts'][0]['text'])
    print()
```
