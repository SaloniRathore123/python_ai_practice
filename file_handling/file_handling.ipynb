```python
# Read the document
with open("meeting.text", "r", encoding="utf-8") as f:
    document = f.read()
    print(document)
```

```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()

# put it in the prompt and ask

prompt = f'''
Here is a meeting transcript:
{document}
Summarize it in 3 short bullet points.
'''

try:
    
    response = client.models.generate_content(
        model="gemini-3.6-flash",  # Switched to an alternative Flash model
        contents=prompt,
    )
    summary = response.text
except Exception as e:
    summary = f"Sorry, something went wrong. Please try again later. ({e})"

print(summary)
```

```python
with open("summary.text", "w", encoding = "utf-8") as f:
    f.write(summary)
print("Summary saved in the summary.text file.")
```
