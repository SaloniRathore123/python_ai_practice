```python
revenue = 0
expense = 10
try:
    margin = profit * 100 / revenue
    print(f"Margin: {margin:.2f}%")
except ZeroDivisionError:
    print("Revenue is 0 so can't calculate the margin. Skipping")
```

```python
# People write this kind of safe function
def safe_margin(revenue, margin):
    try:
        return (revenue - expense) * 100 / revenue
    except ZeroDivisionError:
        return None # signal couldn't compute
print(safe_margin(50, 10))
print(safe_margin(0, 10))
```
