#### What is OOP?
##### Object-Oriented Programming (OOP) in Python is a programming paradigm that structures code by bundling related data and behaviors into reusable units called objects.

#### What is a class?
##### A class is a blueprint or template that defines the structure and capabilities an object will have. You create one using the class keyword.

#### What is an object(instance)?
##### An object is a specific, concrete item created from a class. For example, if Car is the class, a red Toyota and a blue Honda are individual objects.

#### What is an attribute?
##### Attributes are variables inside a class that store data or describe the state of an object (e.g., color or speed).

#### What is a method?
##### Methods are functions defined inside a class that describe the actions or behaviors an object can perform (e.g., drive() or brake()).

#### What is an __init__ and self?
##### The __init__ method is a special constructor that runs automatically when a new object is created to set up its initial attributes. The self parameter refers to the specific object instance being created or used.

#### What is a constructor?
##### A constructor is a special function or method in class-based object-oriented programming that is automatically called when a new object (an instance) is created from a class.

```python
# Employee 1
name1 = "Mohan"
salary1 = 80000

# Employee 2
name2 = "Fatima"
salary2 = 95000

def calculate_bonus(salary):
    return salary * 0.10

print(f"{name1}'s bonus: {calculate_bonus(salary1)}")
print(f"{name2}'s bonus: {calculate_bonus(salary2)}")
```

##### We can bundle data and behaviour into a single entity, and that entity is called a class.
##### A class has two things: properties and methods.
##### emp1 is one instance of the class, and emp2 is another instance of the class.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def calculate_bonus(self):
        return self. salary * 0.10
emp1 = Employee("Mohan", 80000)
emp2 = Employee("Fatima", 95000)
emp1.salary
```

```python
emp1.calculate_bonus()
```

```python
emp2.calculate_bonus()
```

```python
from datetime import date
class Human:
    def __init__(self, name, birth_year):
        self.name = name
        self.birth_year = birth_year
    
    def get_age(self):
        current_year = date.today().year
        return current_year - self.birth_year

    def introduce(self):
        print(f"Hi, my name is {self.name}. I am {self.get_age()} years old.")

saloni = Human("Saloni", 2003)
saloni.get_age()
```

```python
sahil = Human("Sahil", 2005)
sahil.introduce()
```

```python
from pydantic import BaseModel

class Recipe(BaseModel):
    title: str
    calories: int

dal = Recipe(title = "Yellow Dal", calories = 280)
dal.title
```
